<mark>演習 3</mark> `tokyo-power.csv`は2015年度から2023年度までの月別最大電力(東京電力管内)のデータである。(単位:万kW) このデータの階差の分散や自己相関をプロットすることで、この時系列データの周期を調べなさい。

⚠️ 横軸は単純に2015年4月から数えた月数で良い。


```python
%matplotlib inline
import matplotlib.pyplot as plt
import numpy as np
import csv

f = open('tokyo-power.csv', 'r')

data = csv.reader(f)
header = next(f)

# データの表示
power = []

for line in data:
    power.append( float(line[1]) )

f.close()

plt.plot(power)
```
