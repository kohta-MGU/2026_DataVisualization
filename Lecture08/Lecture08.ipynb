Pandasによるデータ解析
-------------------

pandasとよばれるPythonのモジュールについて解説します。Pandasはデータフレームというオブジェクトに対して様々な操作を行うことでデータ解析を行います。データフレームはExcelなどの表計算ソフトの表に相当するものですが、表計算ソフトでは扱いづらいような巨大なデータに対してもPythonのプログラムを組み合わせることで、自動化された処理を行うことが可能です。

また、データフレームのオブジェクト自体に統計処理やグラフ化(可視化)するためのメソッドが多数備わっており、これらを用いることで非常に短いプログラムだけで高度なデータ解析を行うことができます。


データフレーム
------------

Pandasモジュールを用いるには初めにライブラリを

```python
import pandas as pd
```

として読み込みます。Pandasの基本となるデータフレーム(Excelの表の仮想的なもの)は、リストから

```python
data = pd.DataFrame([12.2, 14.8, 11.2, 5.6, 8.6])
```

とすることで作成できます。

データフレームを表示させてみると、

| |0|
|--|--|
|0|12.2|
|1|14.8|
|2|11.2|
|3|5.6|
|4|8.6|

といった表になっていることがわかります。ここで、一番左にある0,1,2といった数字はデータフレーム作成時に自動的に付与されるインデックスで、データの場所を指定するために用いられます。

多数の列がある場合には、列の名前は辞書形式を使って指定することができます。

```python
data = pd.DataFrame({'year':[2004,2005,2006,2007,2008],
                     'value':[12.2, 14.8, 11.2, 5.6, 8.6]})
```

特定の列を取り出すには列の名前を指定し

```python
data['year']
```

とします。また、スライスを用いて特定の行の範囲だけを取り出すこともできます。

```python
data[1:3]
```

データフレームはオブジェクトなので様々なメソッドを持っています。
`data.sum()`、`data.mean()`、`data.std()`は、
それぞれ各列の合計、平均、標準偏差を与えます。統計関係の結果は

```python
data.describe()
```

でまとめて表示させることもできます。

<mark>練習1</mark> あるテストの得点データ

```python
score = [80, 92, 81, 65, 95, 85, 85, 90, 73, 91, 87, 90, 87, 87, 84, 90, 91, 76, 67, 61, 75, 97, 86, 87, 91, 92, 65, 80, 94, 73, 78, 92, 83, 90, 78, 84, 86, 75, 87, 69, 86, 89, 76, 86, 86, 78, 82, 66, 70, 80, 81, 81, 74, 71, 59, 89, 61, 58, 96, 76, 58, 83, 85, 81, 53, 67, 78, 75, 80, 59, 88, 82, 60, 87, 77, 90, 74, 78, 84, 88, 93, 89, 82, 88, 83, 71, 59, 80, 79, 86, 81, 83, 85, 64, 84, 86, 90, 82, 81, 72, 64, 88]
```

を`'score'`列とするデータフレームを作成し、各種統計データを表示しなさい。


````python
import pandas as pd

score = [80, 92, 81, 65, 95, 85, 85, 90, 73, 91,
        87, 90, 87, 87, 84, 90, 91, 76, 67, 61,
        75, 97, 86, 87, 91, 92, 65, 80, 94, 73,
        78, 92, 83, 90, 78, 84, 86, 75, 87, 69,
        86, 89, 76, 86, 86, 78, 82, 66, 70, 80,
        81, 81, 74, 71, 59, 89, 61, 58, 96, 76,
        58, 83, 85, 81, 53, 67, 78, 75, 80, 59,
        88, 82, 60, 87, 77, 90, 74, 78, 84, 88,
        93, 89, 82, 88, 83, 71, 59, 80, 79, 86,
        81, 83, 85, 64, 84, 86, 90, 82, 81, 72,
        64, 88]

df = pd.DataFrame({'score': score})

# 合計
print(df.sum())

# 平均
print(df.mean())

# 標準偏差
print(df.std())

# 統計データ一覧
print(df.describe())
````

各列に対して計算を行い、新しい列を作成するには

```python
data['平成'] = data['year'] - 1988
```

とします。また、条件にあったデータを置き換えるには、

```python
data.loc[ data['value'] <= 6, 'value' ] = 'small'
```

のようにします。(.locで「行」と「列」を指定する。)　ここでは、列`'value'`の値が６以下のものの行に対し、`'value'`列の値を`'small'`という文字列に置き換えています。

<mark>練習3</mark> 上の得点データに対して、S,A,B,C,D評価を行った`'grade'`列を作成するプログラムを作成しなさい。

````python
import pandas as pd

score = [80, 92, 81, 65, 95, 85, 85, 90, 73, 91,
        87, 90, 87, 87, 84, 90, 91, 76, 67, 61,
        75, 97, 86, 87, 91, 92, 65, 80, 94, 73,
        78, 92, 83, 90, 78, 84, 86, 75, 87, 69,
        86, 89, 76, 86, 86, 78, 82, 66, 70, 80,
        81, 81, 74, 71, 59, 89, 61, 58, 96, 76,
        58, 83, 85, 81, 53, 67, 78, 75, 80, 59,
        88, 82, 60, 87, 77, 90, 74, 78, 84, 88,
        93, 89, 82, 88, 83, 71, 59, 80, 79, 86,
        81, 83, 85, 64, 84, 86, 90, 82, 81, 72,
        64, 88]

df = pd.DataFrame({'score': score})

# とりあえず、空のgrade列を作る
df['grade'] = ''

df.loc[ df['score'] >= 90, 'grade' ] ='S'
df.loc[ (90 > df['score']) & (df['score'] >= 80), 'grade' ] ='A'
df.loc[ (80 > df['score']) & (df['score'] >= 70), 'grade' ] ='B'
df.loc[ (70 > df['score']) & (df['score'] >= 60), 'grade' ] ='C'
df.loc[ 60 > df['score'], 'grade' ] ='D'

print(df)
````

PandasによるCSVデータの取り扱い
----------------------------

Pandasモジュールを用いてCSVファイルを読み込むと、自動的にデータフレーム型式に変換して読み込み、その後の取り扱いが容易になります。ここでは、以前用いたCSVデータをpandasで読み込む方法について説明します。

まず、Pandasを用いて`tokyo-temp.csv`から西暦と気温のデータを読み込むには次のようにします。
```python
%matplotlib inline
import pandas as pd

data = pd.read_csv('tokyo-temp.csv')
```

<mark>練習5</mark> pandasデータフレームのplotメソッドを使って、`tokyo-temp.csv`ファイルから読み込んだデータフレームをグラフ表示するプログラムを作成しなさい。

````python
%matplotlib inline
import pandas as pd

df = pd.read_csv('tokyo-temp.csv')

# dfのデータをプロットする(x軸とy軸のデータ列を指定する)
df.plot(x='year', y='temp')
````

第１列をはじめからx軸(インデックス)として扱うには、以下のようにします。

````python
%matplotlib inline
import pandas as pd

df = pd.read_csv('tokyo-temp.csv', index_col=0)

# dfのデータをプロットする(x軸とy軸のデータ列を指定する)
df.plot()
````

Pandasのデータフレームには様々な統計的処理を行うメソッドが用意されており、例えば、`tokyo-temp.csv`ファイルからデータを読み込み、そのデータに対して移動平均を計算して表示するプログラムは次のようになります。

````python
%matplotlib inline
import pandas as pd

df = pd.read_csv('tokyo-temp.csv', index_col=0)

# データから5個(5年)ずつずらしながら取り出し、逐次平均をとる
# sma = Simple Moving Average
sma = df.rolling(5).mean()

# SMA列を新たに作り、移動平均のデータを代入

df['SMA'] = sma

df.plot()
````

CSVで読み込んだ多くのデータを含むデータフレームに関しても、統計関係のメソッドを用いて総和、平均、標準偏差といった様々な統計量を即座に計算することができます。特に、相関係数に関してはデータの解析をする上で非常に使用頻度が高いので、データフレームの`corr`メソッドで簡単に求めることができるようになっています。

pandasを使った相関係数の計算例:

````python
import pandas as pd

data = pd.read_csv('height-weight.csv')

data.corr()
````

Pandasのモジュールを使った場合には相関係数は次のような表(行列)として出力されます。

|      |height|weight|
|------|------|------|
|height|1.000000|0.886877|
|weight|0.886877|1.000000|

この表は行と列のデータそれぞれの組み合わせに対する相関係数を表します。表の対角部分については同じデータの組み合わせに対する相関係数なので、相関係数は必ず1となります。また、表の対角部分を挟んで反対側にある値は同じデータの組み合わせに対する相関係数なので同じ値となります。(上の表では、height-weightの相関係数とweight-heightの相関係数は同じ値です。)
