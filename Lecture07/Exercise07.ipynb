<mark>発展問題</mark> 人を円柱状の物体と仮定すると、人の体積は体の半径を$r$、身長を$h$とすると体積は
$$
V = \pi r^2 \times h
$$
なので、体重$w$は人の密度を$\rho$として、
$$
w = \rho V = \rho \pi r^2 h
$$
となる。

練習3の回帰分析の結果から、$r$は$h$の関数としてどのようなものだと考えられるか？ また、$y=2\pi \, r(h)$としてプロットした結果もあわせて考察しなさい。(ただし、$\rho=10^{-3}\,[{\rm kg}/{\rm cm}^3]$とする。)
