<mark>演習1</mark> $256\times 256$のピクセルサイズの白黒画像を作成し、半径100ピクセルの白色の円を中心に表示させなさい。

ヒント：
- numpyを使って、$256\times 256$のサイズの行列(値0)を作成する。(`img = np.zeros((256, 256))`)
- 行列のインデックス`(i,j)`に対し、
  ```python
  (i-128)**2 + (j-128)**2 <= 100**2
  ```
  であれば行列の値を255とする。
- `plt.imshow(img, cmap='gray')`で表示させる。
