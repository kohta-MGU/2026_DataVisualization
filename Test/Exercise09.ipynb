<mark>演習1</mark> `penguins.csv`データセットを用いて、Adelie(アデリーペンギン)、Chinstrap(ヒゲペンギン)、Gentoo(ジェンツーペンギン)の3種のペンギンを区別する指標について、分析と考察を行いなさい。

![species](lter_penguins.png)

データセットの各項目の意味は以下のとおりです。
- species: 種(Adelie, Chinstrap, Gentoo)
- island: 生息地(Torgersen, Biscoe, Dream)
- bill_length_mm: ペンギンのくちばしの長さ(単位：mm)
- bill_depth_mm： ペンギンのくちばしの奥行き(単位：mm)
- flipper_length_mm： ペンギンの翼の長さ(単位：mm)
- body_mass_g： ペンギンの体重(単位：g)
- gender： ペンギンの性別
- year： データが収集された年

![species](culmen_depth.png)

ヒント：
- 各種データの統計情報を求める
- ペアプロットを作成する
- ペアプロットから種別ごとのクラスターを見つける
- クラスターが明確な指標に対して、KDE分布も含めたジョイントプロット図も作成してみる
- 他にもオスとメスに関する差異などについても考察する(pandasのデータフレームからのデータ抽出などの作業が必要です)
