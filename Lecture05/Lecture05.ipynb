NumPyについて
-----------

NumPyは、プログラミング言語Pythonにおいて数値計算を効率的に行うための拡張モジュールである。効率的な数値計算を行うための型付きの多次元配列（例えばベクトルや行列などを表現できる）のサポートをPythonに加えるとともに、それらを操作するための大規模な高水準の数学関数ライブラリを提供する。(Wikipediaより。)

NumPyではベクトルを以下のように定義します。(縦ベクトルですが、横に表示されます。)

````python
import numpy as np

x = np.array([1, 2, 3])
y = np.array([6, 4, 2])

# ベクトルの和
print("x+y=", x+y)

#　ベクトルの内積
print("x・y=", np.dot(x,y))
````

行列

````python
import numpy as np

A = np.array([[0, 1, 2], [3, 4, 5]])

print(A)
````

行列の演算

````python
import numpy as np

A = np.array([[2, 1, 1], [-1, 2, 1], [1, 1, 0]])
x = np.array([1, 1, 1])

# 行列とベクトルの積
print( "A x =", np.matmul(A,x) )

B = np.array([[0, 1, 1], [0, 0, 1], [0, 0, 0]])

# 行列との積
print( "A B =", np.matmul(A,B) )

# @演算子を使っても行列の積が計算できる
print( "A B =", A @ B )

#  行列式
print( "det=", np.linalg.det(A) )

# 逆行列
print( "A^{-1}=", np.linalg.inv(A) )

# 逆行列であることの確認
print( "A A^{-1} =", A @ np.linalg.inv(A) )

# 固有値と固有ベクトル
eigenvalues, eigenvectors = np.linalg.eig(A)

print("eigenvalues = ", eigenvalues)
print("eigenvectors = ", eigenvectors)
````

Pythonのデフォルトのままでは`range`関数が1刻みの数列しか生成できないので、この様な方法で増分が細かいリストを生成しますが、NumPyモジュールを使えばより簡単にある範囲の数値をより細かい刻みで生成することができます。

乱数
---

乱数(または乱数列)とは今まで得られている数(数列)から次に与えられる数が全く予測できないものを言います。例えば、サイコロを振ったとき、1から6までの目のうち、次にどの目が出るのかは予測することはできません。サイコロを振ったときの目を順に記録していった数列はは乱数列となります。

コンピュータのプログラムにおいて、乱数があると毎回同じ結果とならないような現実の出来事をシミュレートするのに用いることができます。また、統計的な予測を再現するときにも乱数は重要です。

しかし、乱数を生成するようなプログラムを作成しようと思っても、コンピュータは毎回正確に同じ計算を繰り返すだけですので、完全に予測不可能な乱数を生成することは不可能です。とはいえ、実用上問題のない範囲で、値を予測するのが非常に難しい数列を生成するアルゴリズムが多く考案されています。このようなアルゴリズムによって生成された数列を擬似乱数といいます。また、擬似乱数を生成するときに、プログラムを実行している時間やコンピュータに組み込まれたチップの熱的揺らぎなどを組み入れて、より予測が難しく完全な乱数に近いものを生成する工夫がされている場合もあります。

ここでは擬似乱数を生成するためのアルゴリズムについて解説はしませんが、Pythonの標準モジュールの中にすでに擬似乱数を生成する関数がありますので、その関数の利用方法について説明します。

擬似乱数を生成する関数を使うには、まずその関数が定義されたモジュールを、
```python
import random
```
によって読み込みます。

このモジュールを読み込むことで、`random.random()`という関数が使えるようになります。この関数は因数を持たず、0.0から1.0まで(ただし、1.0は含まない)の範囲の擬似乱数を浮動小数点数で返します。

<mark>練習2</mark> 乱数が生成されている様子をプログラムで確かめなさい。

````python
import random

for i in range(10):
    x = random.random()
    print(x)
````

実行するたびに浮動小数点数の異なる数列が表示されるはずです。

この関数があれば、定数倍や定数を加えることで任意の範囲の擬似乱数を得ることができますが、randomモジュールには他の便利な関数も用意されています。

ある範囲にある整数の擬似乱数を発生させるには
```python
random.randint(最小値, 最大値)
```
を用います。

この関数(メソッド)を用いて、1から6までのサイコロの目に対応する擬似乱数列を表示するプログラムは次のようになります。

````python
import random

for i in range(10):
    dice = random.randint(1,6)
    print(dice)
````

実際のサイコロでは、全体に対してある目が出る割合はサイコロを振る回数を増やせば増やすほど均等にそれぞれの目に対してその確率$1/6=1.6666\cdots$に近づいていきます。(大数の法則)

この乱数が本当にサイコロを振ったときの目と同じような振る舞い(確率)を持つのか確認するため、次のプログラムを考えてみましょう。

````python
import random
num = [0,0,0,0,0,0]

for i in range(100):
    dice = random.randint(1,6)
    num[dice - 1] += 1

for n in num:
    print(n / 100)
````

このプログラムでは100回の繰り返しの中で、1から6までの範囲の整数の擬似乱数を発生させ、リスト`num`でそれぞれの目が出た回数をカウントしています。出た目の回数を繰り返し(試行)全体の数で割れば、それぞれの目が出た割合を求めることができますが、プログラムの結果は確率的な予測と同じとなるでしょうか?

回数が少ないうちは確率の予想とは異なるかもしれません。プログラムでも実際に1万回、10万回、100万回とサイコロを振ったときの確率がどうなるかみてみましょう。サイコロを振る回数を変えて何度も比較してみますので、サイコロを振る回数を`N`という変数にしておきます。(またはinput関数を用いて`N`の値を入力するようにしても良いでしょう。)

````python
import random

num = [0,0,0,0,0,0]
N = 10000

for i in range(N):
    dice = random.randint(1,6)
    num[dice - 1] += 1

for n in num:
    print(n / N)
````

サイコロを振る回数を増やすに連れて、それぞれの目が出る割合は等しく$1/6$の値に
近づいていく様子がわかるはずです。

ランダムウォーク
------------

0または1の乱数のように確率がそれぞれ1/2となるような2値の乱数(またはコイントスの裏表)を使って、数直線上を乱数の値0ならば左に$+1/2$、1ならば右に$-1/2$だけ動くことを考えましょう。このような動きは乱数によってふらふら動くのでランダムウォーク(酔歩)と呼ばれます。乱数を使って1回移動することを以下ではステップと呼びましょう。

ランダムウォークで100ステップ後の位置を表示するプログラムは次のようになります。

````python
import random

x = 0
for n in range(100):
    if random.random() < 0.5:
        x += 0.5
    else:
        x -= 0.5

print(x)
````

`x`が各ステップでの位置を表し、初期位置を`x=0`とします。右に行くか左に行くか確率1/2の乱数としては、`random.randint(0,1)`でもいいのですが、左右の確率を別々にする場合も考えて0.0から1.0までの乱数を与える`random.random()`を用いて、その値が$1/2$より大きいか小さいかで判断しています。

次に、100ステップ後の位置について、その平均値を求めてみましょう。上のプログラムを繰り返し使ってもいいですが、これから、ランダムウォークで`N`ステップ後の位置を求めることが何度も出てきますので、関数を使って上のプログラムを書き直してみましょう。`N`ステップ後の位置を返す関数を`position(N)`として定義すると、100ステップ後の位置を、1000回繰り返し求めた時の平均値は次のようになります。

````python
import random

def position(N):
    x = 0
    for n in range(N):
        if random.random() < 0.5:
            x += 0.5
        else:
            x -= 0.5
    return x

s = 0
for m in range(1000):
    s += position(100)

print( s / 1000 )
````

どのような値が出力されたでしょうか? プログラムを何度か実行してみても、結果はおよそ0ぐらいの値になるはずです。

このことは確率論からも理解することができます。まず、ステップ数が$N$のとき、右に行った回数を$n$回とすれば、その確率は、
$$p(N,n)=\frac{{}_N C_n}{2^N}$$
となります。ここで、${}_N C_n$は2項係数を表します。
その時の位置は、
$$x=\frac{n}{2}-\frac{N-n}{2} = n - \frac{N}{2}$$
なので、同じことを繰り返した時の位置の平均値(期待値)は
$$\langle x \rangle  = \sum_{n=0}^N (n - \frac{N}{2})\times p(N,n)$$
となりますが、二項係数の性質$\sum_{n=0}^N {}_N C_n = 2^N$と
$\sum_{n=0}^N n{}_N C_n = N2^{N-1}$を用いると、
$$\langle x \rangle = 0$$
と求まります。また、右に進んだものと左に進んだものの数が原点$x=0$を中心に対称になっていることからも理解することができます。

この結果ではステップ数$N$によらずに位置の平均値はゼロとなります。しかし、これはあくまで平均値であるので、全体的な位置の分布も含めてステップ数に完全によっていないということは意味していません。(原点を中心に左右対称に分布が広がっていれば、平均値は0になります。)

ステップ数に応じて到達点の位置の分布がどのようになるのか、以下の練習問題で調べてみましょう。

<mark>練習3</mark> ランダムウォークの軌跡をグラフで多数重ねて表示するプログラムを作成しなさい。

````python
%matplotlib inline
import matplotlib.pyplot as plt
import random

def position_trajectory(N):
    x = 0
    trajectory = [0]
    
    for n in range(N):
      if random.random() < 0.5:
        x += 0.5
      else:
        x -= 0.5
      
      trajectory.append(x)
    
    return trajectory

N = 5 # ステップ数
n = 100 # 試行回数(サンプル数)

for _ in range(n):
  plt.plot(position_trajectory(N))
````


<mark>練習4</mark> 100回動いた時の最終位置を100回繰り返して求め、その位置の分布をヒストグラムとして棒グラフで表しなさい。

````python
%matplotlib inline
import matplotlib.pyplot as plt
import random

def position(N):
    x = 0
    for n in range(N):
        if random.random() < 0.5:
            x += 0.5
        else:
            x -= 0.5
    return x

N = 100 # ステップ数
hist = [0] * (2*N+1) # 到達位置の数だけの要素を持つリストを作成

n = 100 # 試行回数(サンプル数)
for _ in range(n):
    x = position(N)
    hist[int(x*2)+N] += 1

# ヒストグラムの表示
x = [] # x軸の値
for i in range(len(hist)):
    x.append(i/2 - N/2)

plt.bar(x, hist)
````

<mark>練習5</mark> 試行回数(サンプル数)を変化させると分布はどのような形になるでしょうか。

前問で求めたそれぞれの位置にくる確率を、ステップ数が10の時と1000の時とで比較してみましょう。その比較からどういったことがわかるでしょうか?

ステップ数を増やしても位置の平均値は0のままで変わりませんが、ステップ数が増えるにつれて遠くの位置まで到達したものも現れるようになり、全体として位置の分布が「散らばる」ように見えます。

````python
%matplotlib inline
import matplotlib.pyplot as plt
import random

def position(N):
    x = 0
    for n in range(N):
        if random.random() < 0.5:
            x += 0.5
        else:
            x -= 0.5
    return x

N = 1000 # ステップ数
hist = [0] * (2*N+1) # 到達位置の数だけの要素を持つリストを作成

n = 10000 # 試行回数(サンプル数)
for _ in range(n):
    x = position(N)
    hist[int(x*2)+N] += 1

# ヒストグラムの表示
x = [] # x軸の値
for i in range(len(hist)):
    x.append(i/2 - N/2)

plt.bar(x, hist)
plt.xlim(-100, 100)
````

この位置の分布の散らばりの様子は**分散**を用いて表すことができます。

<mark>練習6</mark> ステップ数が100の時の位置について、1000個のデータに対する分散を求めるプログラムを作成しなさい。

````python
import random

n = 1000

def position(N):
    x = 0
    for n in range(N):
      if random.random() < 0.5:
        x += 0.5
      else:
        x -= 0.5
    return x

N = 1000

pos = []
for i in range(n):
    pos.append( position(N) )

# 位置の平均を求める
avg = sum(pos) / len(pos)

# 分散を求める
sigma2 =0
for x in pos:
    sigma2 += (x - avg) ** 2 / n

print(avg)
print(sigma2)
````

<mark>練習7</mark> ステップ数が10, 100, 1000それぞれについて、1000個のデータの分散を求め、ステップ数と分散の値の関係について考察しなさい。

````python
import random

n = 1000

def position(N):
    x = 0
    for n in range(N):
      if random.random() < 0.5:
        x += 0.5
      else:
        x -= 0.5
    return x

steps = [10, 100, 1000]

for N in steps:
    pos = []
    for i in range(n):
        pos.append( position(N) )

    # 位置の平均を求める
    avg = sum(pos) / len(pos)

    # 分散を求める
    sigma2 =0
    for x in pos:
        sigma2 += (x - avg) ** 2 / n

    print("N=", N)
    print(avg)
    print(sigma2)
````

この結果から、データの散らばり具合がステップ数に応じて変化する様子が、分散という量を使って、感覚ではなく量的にわかるようになりました。このように、分散はデータが平均値に対してどれくらい散らばっているのかを表すのに重要な量となっています。
