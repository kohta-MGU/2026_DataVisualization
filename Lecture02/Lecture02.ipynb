データの種類
---------

- ある集団(集合)に対して、調査や実験の対象の特性を表すものを**変量**(variate)という
- 変量のことを**変数**(variable)ということもある
- ある変量の測定値や観測値、調査結果などの集合を**データ**という
- 変量(変数)には量的なものと質的なものがある
  - 量的変数: 数量で表すことができるもの (例: 長さ、重さ、温度、・・・)
  - 質的変数: 数量で表すことが困難であるもの (例: 性別、職業、既婚/未婚、・・・)
  - ただし、質的変数もデータ処理や解析の際に数値化して量的変数として扱う場合がある

### データの代表値

- データ全体の特徴を適当な1つの数値で表したものを**代表値**という
- データの代表値には、平均値、最頻値、中央値などがある
- データの分布によって、適切な代表値は異なる

#### 平均値

- 平均値: 数値の合計を要素の数で割ったもの  
- データの大きさ(長さ): 要素の数

<mark>例題 1</mark> 次のデータはある学生の5種類のテストの得点である
```python
score = [72, 65, 88, 45, 76]
```
このデータの大きさと平均値を求めよ


````python
score = [72, 65, 88, 45, 76]

n = 0
s = 0

for data in score:
  n += 1
  s += data

print("大きさ= ", n)
print("平均値= ", s/n)
````

リストに対して、要素の数を与える`len()`関数と、合計を与える`sum()`関数がPythonには組み込み関数として存在しているので、それを用いると以下のようにも書ける。

````python
score = [72, 65, 88, 45, 76]

print("大きさ= ", len(score))
print("平均値= ", sum(score)/len(score))
````

一般に、
$$x_1,x_2,\cdots,x_n$$
というデータがあったとき、平均値 $\bar{x}$ は、
$$\bar{x}=\frac{1}{n}\sum_{i=1}^n x_i$$
で与えられる。

#### 最頻値

- データにおいて、もっとも個数の多い値を、そのデータの**最頻値**または**モード**(mode)という

#### 中央値

- データを値の大きさの順に並べたとき、中央の位置にくる値を**中央値**または**メジアン**(median)という
- データの大きさが偶数の場合は中央の2つの値の平均をとって中央値とする


<mark>例題 2</mark> ある商品の価格について、A市とB市で調査した。

A市のデータは、
```python
price_A = [260, 280, 280, 300, 270]
```
で、B市のデータは、
```python
price_B = [280, 280, 260, 100, 280, 270]
```
であった。

1. A市、B市のデータを小さい順に並べ替えなさい。
2. A市、B市のデータの中央値をそれぞれ求めなさい。


````python
def sort(x):
  n = len(x)
  for i in range(n):
    for j in range(n-1, i, -1):
      if x[j] < x[j-1]:
        x[j], x[j-1] = x[j-1], x[j]
  return x

price_A = [260, 280, 280, 300, 270]
print("price_A -> ", sort(price_A))

price_B = [280, 280, 260, 100, 280, 270]
print("price_B -> ", sort(price_B))
````

元々、Pythonには組み込み関数として`sorted()`という関数があるので、自分で関数を作る必要はない。

````python
price_A = [260, 280, 280, 300, 270]
print("price_A -> ", sorted(price_A))

price_B = [280, 280, 260, 100, 280, 270]
print("price_B -> ", sorted(price_B))
````

あるいは、リストオブジェクトに対する操作(メソッド)として(リストの中身自信を変更してしまう)方法もある。

````python
price_A = [260, 280, 280, 300, 270]
price_A.sort()
print("price_A -> ", price_A)

price_B = [280, 280, 260, 100, 280, 270]
price_B.sort()
print("price_B -> ", price_B)
````

これらを使って中央値を求める関数を作ってみる。

````python
def median(x):
  n = len(x)
  x.sort()
  if n % 2 == 0:
    return (x[n//2-1] + x[n//2]) / 2 # データ数が偶数の場合は中央の２つの値の平均値
  else:
    return x[n//2] # データ数が奇数の場合は中央の値

price_A = [260, 280, 280, 300, 270]
price_B = [280, 280, 260, 100, 280, 270]

print("price_A -> ", median(price_A))
print("price_B -> ", median(price_B))
````

### データの散らばりと四分位数

#### データの範囲

- データの散らばりの度合いを表す値として、データの最大値から最小値を引いた差を、データの**範囲**という


<mark>例題 3</mark> 次のデータは、ある年のA市における月毎の降水日数である。
```python
rain_A = [7, 4, 9, 7, 10, 13, 14, 7, 4, 12, 13, 5]
```
このデータの範囲を求めよ。

````python
rain_A = [7, 4, 9, 7, 10, 13, 14, 7, 4, 12, 13, 5]

max = rain_A[0]
min = rain_A[0]

for data in rain_A[1:]:
  if data > max:
    max = data
  if data < min:
    min = data
    
print("最大値 =", max)
print("最小値 =", min)
print("データの範囲 =", max - min)
````

#### 四分位数

- データを値の大きさの順に並べたとき、4等分する位置にくる値を**四分位数**(quartile)という
- 四分位数は、小さい方から順に第1四分位数、第2四分位数、第3四分位数といい、順に$Q_1$、$Q_2$、$Q_3$で表す
- 第2四分位数は中央値と等しい
- より一般に、データの位置をパーセントで表したものを、**パーセンタイル**(percentile)という

### 度数分布

<mark>例題 4</mark> 次のデータは横浜の2024年4月の日ごとの最高気温のデータである。
```python
yokohama_temp =[
17.1, 18.7, 15.7, 18.4, 16.2, 17.3, 21.7, 21.7, 20.0, 16.3,
18.7, 16.8, 20.7, 24.0, 25.6, 23.0, 24.8, 20.5, 24.3, 24.1,
23.1, 17.9, 20.4, 16.4, 24.8, 24.8, 21.8, 27.7, 25.3, 24.5
]
```
このデータのうち、  

A: 15℃以上18℃未満  
B: 18℃以上21℃未満  
C: 21℃以上24℃未満  
D: 24℃以上27℃未満  
E: 27℃以上   

の日数をそれぞれ求めなさい。

````python
yokohama_temp =[
17.1, 18.7, 15.7, 18.4, 16.2, 17.3, 21.7, 21.7, 20.0, 16.3,
18.7, 16.8, 20.7, 24.0, 25.6, 23.0, 24.8, 20.5, 24.3, 24.1,
23.1, 17.9, 20.4, 16.4, 24.8, 24.8, 21.8, 27.7, 25.3, 24.5
]

A = 0
B = 0
C = 0
D = 0
E = 0

for temp in yokohama_temp:
  if   15.0 <= temp < 18.0:
    A += 1
  elif 18.0 <= temp < 21.0:
    B += 1
  elif 21.0 <= temp < 24.0:
    C += 1
  elif 24.0 <= temp < 27.0:
    D += 1
  else:
    E += 1

print("A:", A)
print("B:", B)
print("C:", C)
print("D:", D)
print("E:", E)
````

- データの散らばりの様子を**分布**という
- データをその値の区間ごとにまとめて表にしたものを**度数分布表**という
- 区切られた各区間を**階級**、区間の幅を**階級の幅**、各階級に含まれる値の個数を**度数**という
- 各階級の真ん中の値を**階級値**という

### 偏差

変量$x$のデータの値を$x_1,\ x_2,\ \cdots,\ x_n$、その平均値を$\bar{x}$とするとき、各値から平均値を引いた差 $x_1-\bar{x},\ x_2-\bar{x},\ \cdots,x_n-\bar{x}$ を、それぞれ平均値からの**偏差**という。

偏差の総和は、
$$(x_1-\bar{x})+(x_2-\bar{x})+\cdots+(x_n-\bar{x})
=(x_1+x_2+\cdots+x_n)-n\bar{x}=0$$
となるから、偏差の平均値も必ず0となり、偏差の平均値では散らばりの度合いを表すことができない。

### 分散、標準偏差

偏差そのものでは、平均からのずれとしてプラスとマイナスの値を取るため、データの散らばり度合いを表すことができない。よって、偏差そのものではなく、偏差の2乗の平均を考える。

$n$個のデータがある時、そのデータと全体の平均の差の2乗の相加平均
$$\sigma^2 = \frac{1}{n}\sum_{i=1}^n (x_i - \bar{x})^2$$
を**分散**という。

⚠️ この分散を用いればデータの散らばりについて調べることができます。ここで、平均からのずれとして単純に偏差の相加平均を取ってしまうと、定義より明らかに0となってしまいますので、2乗を取ってから平均を取ることが重要です。分散は、2乗を取ることで常に正の値(平均からの距離の2乗)の平均値となるのでテータの散らばりを表す量として適当なものになります。

<mark>例題 5</mark> データの分散を求める関数`var(x)`を作成しなさい。

````python
def var(x):
  x_var = sum(x) / len(x) # 平均値
  s = 0 # 偏差の二乗の合計をゼロに初期化
  
  for data in x:
    s += (data - x_var) ** 2 # 偏差の二乗をsに加えて合計する
  
  return s / len(x) # 偏差の二乗の平均値を返す
````

<mark>例題 6</mark> データのリスト
```python
x = [38, 53, 43, 41, 35, 46, 50, 38]
```
に対して、このデータの平均値と分散を求めなさい。

````python
x = [38, 53, 43, 41, 35, 46, 50, 38]

print("平均値=", sum(x) / len(x))
print("分散=", var(x))
````

分散の正の平方根の値を標準偏差といい、$\sigma$ で表す。(上では、分散を$\sigma^2$と書いた。)

ある数の正の平方根は`math`モジュールの`math.sqrt()`関数で求められる。

<mark>例題 7</mark> データの標準偏差を求める関数`sigma(x)`を作成しなさい。

````python
import math

def sigma(x):
  x_var = sum(x) / len(x) # 平均値
  s = 0 # 偏差の二乗の合計をゼロに初期化
  
  for data in x:
    s += (data - x_var) ** 2 # 偏差の二乗をsに加えて合計する
  
  return math.sqrt(s / len(x)) # 標準偏差の値を返す
````

<mark>例題 8</mark> 次のデータ
```python

```
に対して、平均、分散、標準偏差の値を求めなさい。

````python
x = [9, 3, 4, 10, 10, 5, 7, 9, 10, 3]

print("平均値=", sum(x)/len(x))
print("分散=", var(x))
print("標準偏差=", sigma(x))
````

#### 分散と平均値の関係式

分散を表す式は次のように変形できる。
$$\begin{split}
\sigma^2 &= \frac{1}{n}\left\{
  (x_1-\bar{x})^2+(x_2-\bar{x})^2+\cdots+(x_n-\bar{x})^2
\right\}\\
&=\frac{1}{n}\left\{
  (x_1^2+x_2^2+\cdots+x_n^2)-2\bar{x}(x_1+x_2+\cdots+x_n)+n(\bar{x})^2
\right\}\\
&=\frac{1}{n}
  (x_1^2+x_2^2+\cdots+x_n^2)-(\bar{x})^2\\
&=\overline{x^2}-(\bar{x})^2
\end{split}$$

この式を用いて、分散や標準偏差を求めても良い。
