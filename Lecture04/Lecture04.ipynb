可読なテキストファイル データ形式
-------------------------

Pythonに限らず、プログラムで利用するためのデータ形式は様々なものがあります。テキストベースのデータ形式だと、Webページを構成しているHTML言語と親和性の良いXML(Extensible Markup Language)やJavaScriptでよく用いられるJSON(JavaScript Object Notation)などもありますが、これらのデータ形式は複雑で取り扱いがやや難しくなります。(ただし、高度なデータ構造を扱う場合には便利な場合も多いでしょう。) また、外部のデータベースプログラムにPythonからアクセスして利用するという方法もあります。

- CSV (Comma Separated Values)  
  値や項目をカンマ(,)で区切って書いたテキストファイル・データ
- XML (eXtensible Markup Language)  
  HTMLから派生したデータ構造も含む書式
- JSON (JavaScript Object Notation)  
  JavaScriptというプログラミング言語におけるオブジェクトの書き方を参考に作られたデータフォーマット(データの記述形式)
- YAML (Yet Another Markup Language, YAML Ain't a Markup Language)  
  構造化データやオブジェクトを文字列にシリアライズ(直列化)するためのデータ形式

その他、バイナリデータ(可読性なし)ですが、Excelファイル(xlsx)も外部データ形式として使用することができます。また、ExcelでもCSV形式を読み込んだり保存したりできるので、今回は主にCSVデータを取り扱う方法について学びます。

CSVの構造
--------

ここでは、より単純なデータ形式であるCSV(Comma-Separated Values)を扱います。CSVはExcelなどの表計算ソフトと相性がよく、多くの表計算ソフトではデータをCSV形式で出力する機能がありますので、表計算ソフトである程度加工したデータをPythonに受け渡す際にも便利です。

CSVとはその名が表すように、コンマ記号で分割して並べられたデータです。コンマは表計算ソフトでいう列のデータを分け、改行で区切られたデータは表計算ソフトられた各行に対応します。

```
項目1, 項目2, ... , 項目n [改行]
値1 ,   値2, ... , 値n  [改行]
...
```

項目を区切る文字は必ずしもカンマである必要はありません。(タブで区切った場合には、TSVとも呼ばれます。)

CSVデータの読み込み
---------------

CSV形式のファイルは単なるテキストデータですから、ファイルの標準読み込みを使ってデータを読み込み、それを加工(処理)することでPythonのデータとして扱うことができます。

````python
f = open('tokyo-temp.csv', 'r')

for data in f:
    print(data)

f.close()
````

しかし、このままではdataの変数に各行のデータがコンマを含む文字列データとして入力されますので、コンマごとに分かれた数値として扱うためにはプログラムを使ってデータを加工する必要があります。(`x, y = data`などとするとエラーになります。)

### csvモジュール

CSVのデータ形式は決まっていて、毎回同じ処理をPythonのプログラムとして書くのは面倒ですので、Pythonでは標準ライブラリとしてCSV形式のデータを扱うcsvというモジュールが用意されています。

まず、csvモジュールの使い方を学ぶために、`tokyo-temp.csv`というサンプルCSVデータを用います。`tokyo-temp.csv`をExcelなどの表計算ソフトで開いてみると、1行目に「year」と「temp」という項目が書かれ、2行目以降に1976年から2016年までの東京の平均気温が並んだデータとなっていることがわかります。実際の中身は、これらのデータがコンマと改行によって分けられた単なるテキストデータです。

まずは、このCSVファイルを通常のテキストファイルと同じように扱ってみましょう。

csvモジュールは各行のデータをより扱いやすくする方法を与えます。csvモジュールを使うためには、まずcsvモジュールをインポートし、ファイルオブジェクトからデータを読み出す時に、`csv.reader()`関数を用います。`csv.reader()`は`for`文などを使って反復処理ができるオブジェクトを返します。

````python
import csv

f = open('tokyo-temp.csv', 'r')

data = csv.reader(f)

for line in data:
    print(line)

f.close()
````

`csv.reader()`関数を使うと、各行のデータがリストとして読み込まれているのがわかると思います。(ただし、各データの要素は全て**文字列**のままであることに注意します。)

上のプログラムの例では、1行目にあるデータの説明(ヘッダー)も全て他のデータと同じように扱われてしまいます。ヘッダーのデータを読み飛ばすには`next()`関数を用います。ヘッダーなどの実際のデータ以外の部分を読み飛ばせば、以後のデータを変数に代入して直接扱うこともできます。

````python
import csv

f = open('tokyo-temp.csv', 'r')

data = csv.reader(f)
header = next(f) # 1行目をヘッダーとして取り出し、次に進める

year = []
temp = []

for line in data:
    year.append( int(line[0]) )   # 年は整数なのでint()で変換してyearリストに追加
    temp.append( float(line[1]) ) # 気温は浮動小数点数なのでfloat()で変換してtempリストに追加

print(header)
print(year)
print(temp)

f.close()
````

散布図と相関
---------

ある2つのデータ列$(x_i,y_i)$ ($i=1,2,\ldots,N$)に対して、一方の値が増加したときにもう一方が増加する傾向があるとき、それら2つのデータの間には**正の相関**があるといいます。逆に、一方の値が増加したとき他方が減少する傾向があれば**負の相関**があるといいます。また、正の相関も負の相関も見られないときは相関が無いといいます。

例として、あるクラス20人の身長と体重をまとめたデータ`height-weight.csv`を散布図としてプロットして見ましょう。

````python
%matplotlib inline
import matplotlib.pyplot as plt
import csv

f = open('height-weight.csv','r')

data = csv.reader(f)
header = next(f)

height = []
weight = []

for line in data:
    height.append( float(line[0]) )
    weight.append( float(line[1]) )

f.close()

plt.scatter( height, weight )
````

このデータにはどのような相関があると言えるでしょうか?

このようにデータを散布図でプロットしてみると、2つのデータの間に相関があるかどうか、ある程度視覚的に知ることができます。データの間に明らかな相関がある場合にはこのような方法でもいいですが、相関があるかどうか判断が難しいときや、色々なデータの組に対して相関の大きさを比較したい場合にはその相関の度合いを測る量が必要になります。

まず、$N$個のデータの組
$$(x_1,y_1), \quad (x_2,y_2), \quad \cdots \quad, (x_N,y_N)$$
について、$x$、$y$のデータの平均値を、それぞれ$\bar{x}$、$\bar{y}$とします。このデータに対して共分散と呼ばれる量を
$$\sigma_{xy} = \frac{1}{N}
\sum_{i=1}^N
(x_i-\bar{x})(y_i-\bar{y})$$
と定義すると、その値によって相関の大きさを表すことができます。特に、共分散の値が正の時には正の相関、共分散の値が負の時には負の相関があります。
相関が無い場合には共分散はゼロに近い値になります。

例えば、最も相関が明らかな$y_i=a x_i+b$という関係がデータにある場合、共分散を求めてみると、
$$\begin{split}
\sigma_{xy} &= \frac{1}{N}\sum_{i=1}^N(x_i-\bar{x})(y_i-\bar{y})\\
&= a\frac{1}{N}\sum_{i=1}^N(x_i-\bar{x})^2\\
&= a\sigma_x^2
\end{split}$$
となります。($\bar{y}=a\bar{x}+b$となることを用いました。) ここで、$x_i$の分散$\sigma_x^2=\frac{1}{N}\sum_{i=1}^N(x_i-\bar{x})^2$は常に正の値となりますので、$a$が正の値の時は正の相関、負の値の時は負の相関となります。(この例は最も強い相関となります。)

<mark>練習 2</mark> 要素の個数が同じ2つのリスト`x`と`y`が与えられたとき、共分散を返す関数`cov(x,y)`を定義しなさい。

````python
def cov(x, y):
  N = len(x)
  x_bar = sum(x) / N
  y_bar = sum(y) / N
  
  s = 0
  
  for i in range(len(x)):
    s += (x[i] - x_bar)*(y[i] - y_bar)
  
  return s / N
````

<mark>練習 3</mark> `tokyo-temp.csv`のデータに対して共分散を求めるプログラムを作成しましょう。

````python
%matplotlib inline
import matplotlib.pyplot as plt
import csv

def cov(x, y):
  N = len(x)
  x_bar = sum(x) / N
  y_bar = sum(y) / N
  
  s = 0
  
  for i in range(len(x)):
    s += (x[i] - x_bar)*(y[i] - y_bar)
  
  return s / N

f = open('tokyo-temp.csv','r')

data = csv.reader(f)
header = next(f)

x = []
y = []

for line in data:
    x.append( float(line[0]) )
    y.append( float(line[1]) )

f.close()

plt.scatter( x, y )

print(cov(x, y))
````

共分散は相関の傾向を表す量になりますが、その値そのものはデータの散らばり具合(個々の分散)によってしまいますので、そのままでは異なるデータの集合同士で比較することはできません。そのため、相関の強さを表す指標として、共分散を$x$、$y$それぞれのデータの標準偏差(分散の平方根)
$$\begin{align}\sigma_x&=&\sqrt{\frac{1}{N}
\sum_{i=1}^N(x_i-\bar{x})^2}\nonumber\\
\sigma_y&=&\sqrt{\frac{1}{N}
\sum_{i=1}^N(y_i-\bar{y})^2}\nonumber
\end{align}$$
で割った**相関係数**$r$を用います。
$$r = \frac{\sigma_{xy}}{\sigma_x \sigma_y}$$

相関係数は$-1\leqq r \leqq 1$の間の値をとり、その大きさの大小で相関の強さを表すことができます。

<mark>練習 4</mark> $y_i=ax_i+b$の時には$a$の正負によって$r=\pm 1$となることを確認しなさい。

一般的には相関係数の値の絶対値が0.7を超える場合、それらのデータの間の相関が十分に強いと判断します。(相関係数が0.7以上であれば正の相関が、-0.7以下であれば負の相関が強いと言えます。)

相関係数はデータの間の関連を評価するのに便利な指標ですが、相関の強さを持って、データの間に因果関係があると結論づけることはできないので注意が必要です。例えば、全く無関係でありながら時間に対して単調増加をする2つのデータに対して相関係数を計算すると強い正の相関($r>0.7$)を示すことがあります。

相関係数によるデータの比較は、データ間の因果関係にある程度の仮説が存在する場合の検証として考えるようにしましょう。

<mark>練習 5</mark> 2つのデータのリスト`x`と`y`に対して、相関係数を返す関数`r(x,y)`を定義しなさい。

````python
import math

# 標準偏差を求める関数
def sigma(x):
  x_var = sum(x) / len(x) # 平均値
  s = 0 # 偏差の二乗の合計をゼロに初期化
  
  for data in x:
    s += (data - x_var) ** 2 # 偏差の二乗をsに加えて合計する
  
  return math.sqrt(s / len(x)) # 標準偏差の値を返す

def cov(x, y):
  N = len(x)
  x_bar = sum(x) / N
  y_bar = sum(y) / N
  
  s = 0
  
  for i in range(len(x)):
    s += (x[i] - x_bar)*(y[i] - y_bar)
  
  return s / N

# 相関係数を求める関数
def r(x, y):
  return cov(x, y) / (sigma(x) * sigma(y))
````
